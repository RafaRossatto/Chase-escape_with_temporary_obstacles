```python
import pandas as pd
import numpy as np
import os

def gerar_xyz_com_elementos(path_hunter, path_prey, pasta_saida='ovito_elementos'):
    """
    Gera arquivos .xyz com elementos químicos para cores:
    - Hunters: N (Nitrogênio) -> Azul no OVITO
    - Prey: O (Oxigênio) -> Vermelho no OVITO
    """
    
    # Criar pasta
    if not os.path.exists(pasta_saida):
        os.makedirs(pasta_saida)
        print(f"Pasta criada: {pasta_saida}")
    
    # Carregar dados
    print("Carregando hunters...")
    df_h = pd.read_csv(path_hunter)
    df_h['elemento'] = 'N'  # Hunters = Nitrogênio (azul)
    
    print("Carregando prey...")
    df_p = pd.read_csv(path_prey)
    df_p['elemento'] = 'O'  # Prey = Oxigênio (vermelho)
    
    # Combinar
    df = pd.concat([df_h, df_p], ignore_index=True)
    df = df.sort_values('timestep')
    
    timesteps = sorted(df['timestep'].unique())
    
    print(f"Total de timesteps: {len(timesteps)}")
    
    # Estatísticas
    total_hunters = len(df[df['elemento'] == 'N'])
    total_prey = len(df[df['elemento'] == 'O'])
    print(f"Total de hunters (N - azul): {total_hunters}")
    print(f"Total de prey (O - vermelho): {total_prey}")
    
    # Para cada timestep
    for i, t in enumerate(timesteps):
        df_t = df[df['timestep'] == t]
        
        # Nome do arquivo
        nome_arquivo = f"frame_{i:06d}.xyz"
        caminho = os.path.join(pasta_saida, nome_arquivo)
        
        # Escrever arquivo
        with open(caminho, 'w') as f:
            # Linha 1: número de partículas
            f.write(f"{len(df_t)}\n")
            
            # Linha 2: comentário
            f.write(f"Timestep {t} - Hunters(N-azul) Prey(O-vermelho)\n")
            
            # Linhas 3+: [elemento] [x] [y] [z]
            for _, row in df_t.iterrows():
                f.write(f"{row['elemento']} {row['x']:.3f} {row['y']:.3f} 0.0\n")
        
        if i % 10 == 0:
            print(f"  Frame {i}: {len(df_t)} partículas")
    
    # Criar arquivo de cores personalizadas (opcional)
    with open(os.path.join(pasta_saida, "cores_personalizadas.xml"), 'w') as f:
        f.write("""<?xml version="1.0" encoding="UTF-8"?>
<ovito>
    <colors>
        <element symbol="N" name="Nitrogen" color="0.2 0.5 1.0"/>  <!-- Azul mais vivo -->
        <element symbol="O" name="Oxygen" color="1.0 0.2 0.2"/>    <!-- Vermelho mais vivo -->
    </colors>
</ovito>
""")
    
    # Instruções
    with open(os.path.join(pasta_saida, "INSTRUCOES.txt"), 'w') as f:
        f.write("""INSTRUÇÕES PARA OVITO
====================

1. Abra o OVITO
2. File -> Load File -> selecione: frame_000000.xyz
3. Clique SIM para carregar sequência

4. Configure a caixa 128x128:
   Edit -> Simulation Cell:
   - X: (128, 0, 0)
   - Y: (0, 128, 0)
   - Z: (0, 0, 1)
   - Origin: (0, 0, 0)

5. Para vista 2D: Tecla '7'

🎨 CORES AUTOMÁTICAS:
- N (Nitrogênio) = AZUL (hunters)
- O (Oxigênio) = VERMELHO (prey)

OPCIONAL: Carregar cores personalizadas
- File -> Import -> Colors -> selecione 'cores_personalizadas.xml'
""")
    
    print(f"\n✅ Arquivos gerados em: {pasta_saida}")
    print(f"   {len(timesteps)} arquivos .xyz criados")
    print(f"\n🎨 Cores:")
    print(f"   - Hunters: N (Nitrogênio) -> AZUL")
    print(f"   - Prey: O (Oxigênio) -> VERMELHO")

# ==============================================
# VERSÃO ALTERNATIVA COM OUTROS ELEMENTOS
# ==============================================

def gerar_xyz_com_outros_elementos(path_hunter, path_prey, pasta_saida='ovito_outros'):
    """
    Outras opções de elementos e cores:
    - Hunters: Fe (Ferro) -> Prata/Acinzentado
    - Prey: Au (Ouro) -> Amarelo
    """
    
    if not os.path.exists(pasta_saida):
        os.makedirs(pasta_saida)
    
    print("Carregando hunters (Fe - prata)...")
    df_h = pd.read_csv(path_hunter)
    df_h['elemento'] = 'Fe'
    
    print("Carregando prey (Au - ouro)...")
    df_p = pd.read_csv(path_prey)
    df_p['elemento'] = 'Au'
    
    df = pd.concat([df_h, df_p], ignore_index=True)
    df = df.sort_values('timestep')
    
    timesteps = sorted(df['timestep'].unique())
    
    for i, t in enumerate(timesteps):
        df_t = df[df['timestep'] == t]
        nome_arquivo = f"frame_{i:06d}.xyz"
        caminho = os.path.join(pasta_saida, nome_arquivo)
        
        with open(caminho, 'w') as f:
            f.write(f"{len(df_t)}\n")
            f.write(f"Timestep {t} - Hunters(Fe) Prey(Au)\n")
            for _, row in df_t.iterrows():
                f.write(f"{row['elemento']} {row['x']:.3f} {row['y']:.3f} 0.0\n")
        
        if i % 10 == 0:
            print(f"  Frame {i}: {len(df_t)} partículas")
    
    print(f"\n✅ Arquivos gerados em: {pasta_saida}")
    print(f"   Cores: Fe (prata) = hunters, Au (ouro) = prey")

# ==============================================
# EXECUTAR
# ==============================================

# SEUS CAMINHOS
path_hunter = "/home/rafarossatto/Chase-escape_with_temporary_obstacles/c++/Np=free*0.25/L_128/Nc=0.5/src/NC_1433_NE_2867_O_4915_TCC_1.000000_SR_2_TCT_1.000000_SR_2.dat_run_50_hunter_trajectories.csv"
path_prey = "/home/rafarossatto/Chase-escape_with_temporary_obstacles/c++/Np=free*0.25/L_128/Nc=0.5/src/NC_1433_NE_2867_O_4915_TCC_1.000000_SR_2_TCT_1.000000_SR_2.dat_run_50_prey_trajectories.csv"

# Verificar arquivos
if not os.path.exists(path_hunter):
    print(f"ERRO: Arquivo não encontrado: {path_hunter}")
elif not os.path.exists(path_prey):
    print(f"ERRO: Arquivo não encontrado: {path_prey}")
else:
    # ESCOLHA UMA OPÇÃO:
    
    # Opção 1: N (azul) para hunters, O (vermelho) para prey
    gerar_xyz_com_elementos(path_hunter, path_prey, pasta_saida='ovito_N_azul_O_vermelho')
    
    # Opção 2: Fe (prata) para hunters, Au (ouro) para prey
    # gerar_xyz_com_outros_elementos(path_hunter, path_prey, pasta_saida='ovito_Fe_Au')
    
    print("\n" + "="*50)
    print("✅ ARQUIVOS PRONTOS!")
    print("="*50)
    print("\nCores automáticas do OVITO:")
    print("  - N (Nitrogênio) → AZUL (hunters)")
    print("  - O (Oxigênio) → VERMELHO (prey)")
    print("\nAbra no OVITO e veja as cores!")
```

## Com obstaculos mais recente



```python
import pandas as pd
import numpy as np
import os

def diagnosticar_arquivo_txt(path_obstaculos):
    """
    Diagnostica o formato do arquivo .txt
    """
    print(f"\n=== DIAGNÓSTICO DO ARQUIVO .TXT (OBSTÁCULOS FIXOS) ===")
    print(f"Arquivo: {path_obstaculos}")
    
    try:
        with open(path_obstaculos, 'r') as f:
            linhas = [f.readline().strip() for _ in range(5)]
        
        print("\n📄 Primeiras 5 linhas:")
        for i, linha in enumerate(linhas, 1):
            print(f"   {i}: '{linha}'")
        
        # Analisar primeira linha
        primeira = linhas[0]
        partes = primeira.split()
        print(f"\n📊 Número de colunas: {len(partes)}")
        
        if len(partes) == 2:
            print("📌 FORMATO: [x, y] (obstáculos fixos)")
            return 'xy_fixo'
        else:
            print("📌 Assumindo formato [x, y] (ignorando colunas extras)")
            return 'xy_fixo'
            
    except Exception as e:
        print(f"❌ Erro ao ler arquivo: {e}")
        return 'erro'

def ler_obstaculos_fixos(path_obstaculos):
    """
    Lê obstáculos que são FIXOS
    """
    print(f"\n=== LENDO OBSTÁCULOS FIXOS ===")
    
    try:
        # Tenta ler apenas as duas primeiras colunas
        df = pd.read_csv(path_obstaculos, delim_whitespace=True, header=None,
                         usecols=[0,1], names=['x', 'y'])
        print(f"✅ Lido: {len(df)} obstáculos fixos")
        print(f"   Range X: {df['x'].min():.1f} a {df['x'].max():.1f}")
        print(f"   Range Y: {df['y'].min():.1f} a {df['y'].max():.1f}")
        return df
    except Exception as e:
        print(f"❌ Erro: {e}")
        return None

def gerar_xyz_completo(path_hunter, path_prey, path_obstaculos, 
                      pasta_saida='ovito_completo'):
    """
    Gera arquivos .xyz com CAIXA e obstáculos fixos
    """
    
    # Criar pasta
    if not os.path.exists(pasta_saida):
        os.makedirs(pasta_saida)
        print(f"Pasta criada: {pasta_saida}")
    
    # Carregar dados
    print("\n=== CARREGANDO ARQUIVOS ===")
    
    # Hunters
    print("Carregando hunters (N)...")
    df_h = pd.read_csv(path_hunter)
    df_h['elemento'] = 'N'
    
    # Prey
    print("Carregando prey (O)...")
    df_p = pd.read_csv(path_prey)
    df_p['elemento'] = 'O'
    
    # Obstáculos FIXOS
    print("Carregando obstáculos (C)...")
    df_o = ler_obstaculos_fixos(path_obstaculos)
    if df_o is None:
        print("❌ Erro nos obstáculos. Abortando.")
        return None
    
    df_o['elemento'] = 'C'
    
    # Determinar timesteps comuns
    timesteps = sorted(set(df_h['timestep'].unique()) & set(df_p['timestep'].unique()))
    
    print(f"\n=== CONFIGURAÇÃO ===")
    print(f"CAIXA: 128 x 128")
    print(f"Timesteps: {len(timesteps)}")
    print(f"Hunters (N): {len(df_h)}")
    print(f"Prey (O): {len(df_p)}")
    print(f"Obstáculos fixos (C): {len(df_o)}")
    
    # Gerar frames
    print(f"\n=== GERANDO {len(timesteps)} FRAMES ===")
    
    for i, t in enumerate(timesteps):
        # Filtrar dados móveis
        df_h_t = df_h[df_h['timestep'] == t].copy()
        df_p_t = df_p[df_p['timestep'] == t].copy()
        
        # Adicionar obstáculos fixos (cópia)
        df_o_t = df_o.copy()
        
        # Combinar
        df_t = pd.concat([df_h_t, df_p_t, df_o_t], ignore_index=True)
        
        # Nome do arquivo
        nome_arquivo = f"frame_{i:06d}.xyz"
        caminho = os.path.join(pasta_saida, nome_arquivo)
        
        # ESCREVER ARQUIVO - FORMATO CORRETO!
        with open(caminho, 'w') as f:
            # Linha 1: número de partículas
            f.write(f"{len(df_t)}\n")
            
            # Linha 2: COMENTÁRIO COM A CAIXA!
            f.write(f"Timestep {t} | Box: 0-128 x 0-128 | N={len(df_h_t)} O={len(df_p_t)} C={len(df_o_t)}\n")
            
            # Linhas 3+: dados
            for _, row in df_t.iterrows():
                f.write(f"{row['elemento']} {row['x']:.3f} {row['y']:.3f} 0.0\n")
        
        if i % 10 == 0:
            print(f"  Frame {i:3d}: {len(df_t):3d} partículas "
                  f"(N={len(df_h_t):3d}, O={len(df_p_t):3d}, C={len(df_o_t):3d})")
    
    # INSTRUÇÕES DETALHADAS
    with open(os.path.join(pasta_saida, "INSTRUCOES.txt"), 'w') as f:
        f.write("""INSTRUÇÕES PARA OVITO - VERSÃO COMPLETA
=====================================

1. Abra o OVITO
2. File -> Load File -> selecione: frame_000000.xyz
3. Clique SIM para carregar sequência

4. CONFIGURAR A CAIXA (OBRIGATÓRIO):
   Edit -> Simulation Cell
   
   Cell vectors:
   ╔════════════════════════════════╗
   ║ X: (128, 0, 0)                ║
   ║ Y: (0, 128, 0)                ║
   ║ Z: (0, 0, 1)                  ║
   ║ Origin: (0, 0, 0)             ║
   ╚════════════════════════════════╝

5. Visualização:
   - Tecla '7' = vista superior (2D)
   - Zoom: roda do mouse
   - Mover: botão direito

6. Cores automáticas (elementos):
   - N (Nitrogênio) = AZUL (hunters)
   - O (Oxigênio) = VERMELHO (prey)
   - C (Carbono) = CINZA (obstáculos fixos)

7. Para ajustar cores:
   - Clique com botão direito no elemento
   - Edit -> Color

📊 ESTATÍSTICAS DOS ARQUIVOS:
""")
        # Adicionar estatísticas
        for i, t in enumerate(timesteps[:10]):
            df_h_t = df_h[df_h['timestep'] == t]
            df_p_t = df_p[df_p['timestep'] == t]
            f.write(f"Frame {i:3d} (t={t:3d}): H={len(df_h_t):3d} P={len(df_p_t):3d} O={len(df_o):3d}\n")
        
        f.write(f"\nTotal de frames: {len(timesteps)}")
        f.write(f"\nObstáculos fixos: {len(df_o)} pontos")
    
    print(f"\n✅ ARQUIVOS PRONTOS!")
    print(f"   Pasta: {pasta_saida}")
    print(f"   Frames: {len(timesteps)}")
    print(f"\n👉 NÃO ESQUEÇA DE CONFIGURAR A CAIXA NO OVITO:")
    print("   Edit -> Simulation Cell -> (128,0,0) (0,128,0) (0,0,1)")

# ==============================================
# EXECUTAR
# ==============================================

if __name__ == "__main__":
    
    # SEUS CAMINHOS
    path_hunter = "/home/rafarossatto/Chase-escape_with_temporary_obstacles/c++/Np=free*0.25/L_128/Nc=0.5/src/NC_1433_NE_2867_O_4915_TCC_1.000000_SR_2_TCT_1.000000_SR_2.dat_run_50_hunter_trajectories.csv"
    path_prey = "/home/rafarossatto/Chase-escape_with_temporary_obstacles/c++/Np=free*0.25/L_128/Nc=0.5/src/NC_1433_NE_2867_O_4915_TCC_1.000000_SR_2_TCT_1.000000_SR_2.dat_run_50_prey_trajectories.csv"
    path_obstaculos = "/home/rafarossatto/Chase-escape_with_temporary_obstacles/c++/Np=free*0.25/L_128/Nc=0.5/obs_4915/nC_1433/run_50/obstacules.txt"  # <-- COLOQUE O CAMINHO DO SEU ARQUIVO AQUI
    
    # Verificar arquivos
    print("=== VERIFICANDO ARQUIVOS ===")
    for path in [path_hunter, path_prey, path_obstaculos]:
        if os.path.exists(path):
            print(f"✓ {os.path.basename(path)}")
        else:
            print(f"✗ Arquivo não encontrado: {path}")
    
    # Gerar arquivos
    gerar_xyz_completo(
        path_hunter, 
        path_prey, 
        path_obstaculos,
        pasta_saida='ovito_final_com_caixa'
    )
    
    print("\n" + "="*60)
    print("✅ PROCESSO CONCLUÍDO!")
    print("="*60)
    print("\n📁 Pasta criada: ovito_final_com_caixa")
    print("\n📌 LEMBRE-SE: Configurar a caixa manualmente no OVITO:")
    print("   Edit -> Simulation Cell")
    print("   X: (128, 0, 0)")
    print("   Y: (0, 128, 0)")
    print("   Z: (0, 0, 1)")
    print("   Origin: (0, 0, 0)")
```
